```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import seaborn as sbn

data = pd.read_excel(r'/workspaces/Series_Temporais/Data/AirQualityUCI.xlsx')
```

```python
display(data)
```

```python
plt.figure(figsize = (15, 6),dpi = 80)
plt.plot(data.iloc[:,3])
```

```python
media = np.mean(data.iloc[:,3])
desvio_padrao = np.std(data.iloc[:,3])

print(f"Média: {media} \nDesvio padrão: {desvio_padrao}")
```

```python
from statsmodels.tsa.seasonal import seasonal_decompose

decompose_result = seasonal_decompose(data.iloc[:,3], model = "additive", period = 12)

decompose_result.plot()
```
